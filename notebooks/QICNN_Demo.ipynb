# QICNN Deepfake Detection – Demo Notebook

This notebook walks through:
1. Building the QICNN model
2. Running a forward pass on a synthetic image
3. Demonstrating quantum feature encoding
4. Computing evaluation metrics

```python
import sys
sys.path.insert(0, '../src')

import torch
import pennylane as qml
from model.qicnn import QICNN
from quantum.quantum_preprocessing import QuantumFeatureMap, AngleEncoder
from utils.metrics import compute_metrics
```

## 1. Build a Small QICNN Model

```python
model = QICNN(
    num_classes=2,
    image_size=64,
    channels=[16, 32],
    n_qubits=4,
    n_quantum_layers=1,
    encoding='angle',
)
print(f'Parameters: {sum(p.numel() for p in model.parameters()):,}')
```

## 2. Forward Pass on Synthetic Images

```python
batch = torch.randn(4, 3, 64, 64)
logits = model(batch)
probs = torch.softmax(logits, dim=1)
print('Logits shape:', logits.shape)
print('Probabilities (fake):', probs[:, 1].detach().numpy().round(3))
```

## 3. Quantum Feature Encoding

```python
enc = AngleEncoder(n_qubits=4)
features = torch.randn(8)
q_out = enc.encode(features)
print('Quantum expectation values:', q_out.detach().numpy().round(4))
```

## 4. Evaluation Metrics

```python
import torch
# Simulate logits and labels
logits = torch.tensor([[3.0, -1.0], [-2.0, 2.0], [1.5, -0.5], [-0.3, 1.8]])
labels = torch.tensor([0, 1, 0, 1])
metrics = compute_metrics(logits, labels)
for k, v in metrics.items():
    print(f'{k}: {v:.4f}')
```
